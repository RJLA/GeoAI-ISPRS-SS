## Batches

A batch is a subset of the training data. Instead of using the entire dataset to
update the model's parameters, the data is divided into smaller chunks called
batches.

#### Why use batches?

`Efficiency`:

- Memory Management: Handling the entire dataset at once can be computationally expensive and require a lot of memory. Batches allow processing smaller chunks, making it feasible to train large models.
- Speed: Training with batches can be faster than using the entire dataset, as smaller subsets of data can be processed more quickly.

`Stability`:

Gradient Estimation: Using batches helps provide a more accurate estimate of the gradient, which can lead to more stable and effective training compared to using a single data point at a time.

`Parallelism`:

Optimization: Modern hardware like GPUs can efficiently handle batch processing,
allowing for parallel computations and faster training times.



`Batch Size`: The number of simultaneous data points used in a single iteration
of training. Common batch sizes include 32, 64, 128, and 256, but the optimal
size can vary depending on the dataset and model. 
- Defines the speed and stability of the training process.
  - Usual batch sizes are 1 - 512
  - Good default is 32 or 64
