```python
import pandas as pd
import pickle
import numpy as np
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import MinMaxScaler
```

```python
X_train = pd.read_csv("csv_files/X_train.csv")
X_test = pd.read_csv("csv_files/X_test.csv")
y_train = pd.read_csv("csv_files/y_train_encoded.csv")
y_test = pd.read_csv("csv_files/y_test_encoded.csv")
X_train.head()
```

```python
# Subset the data to only include the bands used in fitting the model
rgbnir_train = X_train[["BLUE", "GREEN", "RED"]]
rgbnir_test = X_test[["BLUE", "GREEN", "RED"]]
```

## Pipeline in Scikit-learn

A Pipeline in scikit-learn is a powerful tool for automating the workflow of machine learning models. It sequentially combines multiple steps, such as data preprocessing, feature selection, and model training, into a single object. This streamlined process ensures that the same transformations are applied consistently during both training and prediction.

### Benefits of Using Pipelines

1. **Simplifies Workflow**: Combines multiple preprocessing steps and model training into a single coherent process, reducing the complexity of the code.
2. **Consistency**: Ensures that the same preprocessing steps are applied during both training and prediction, preventing data leakage and ensuring consistent data transformations.
3. **Readability and Maintainability**: Enhances the readability and maintainability of the code by organizing the workflow in a clear, logical manner.

### How to Create a Pipeline

A typical pipeline in scikit-learn consists of the following steps:

1. **Preprocessing**: Includes steps such as scaling, encoding categorical variables, etc.
2. **Feature Selection**: Involves selecting the most relevant features for the model.
3. **Model Training**: The final step where the model is trained on the preprocessed data.

```python
# Define the pipeline
min_max_scaler = MinMaxScaler() # Instantiate MinMaxScaler
lr = LogisticRegression(solver='lbfgs', max_iter=1000, random_state=1) # Instantiate Logistic Regression

pipeline = Pipeline([
    ('min_max_scaler', min_max_scaler),
    ('lr', lr)
])
```

```python
# save the model using pickle
# merge the train and test datasets

X_all = pd.concat([rgbnir_train, rgbnir_test])
y_all = pd.concat([y_train, y_test])

# train the model using the whole dataset
pipeline.fit(X_all, y_all)
with open('trained_models/simple_pipeline.pkl', 'wb') as file:
    pickle.dump(pipeline, file)
```
