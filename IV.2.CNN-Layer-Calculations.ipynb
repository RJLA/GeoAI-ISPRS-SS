| Dimension    | Structure                                | Used for                               | Example |
|-----------|-----------------------------------|-----------------------------------------------|--------------- |
| 1  | batch size                         | Labels/Predictions   | [16]
| 2 | batch size and feature               | nn.Linear()           | [16, 512]
| 3    | batch size, channels, feature               | nn.Conv1d()           | [16, 1, 512]
| 4  | batch size, channels, height, width | nn.Conv2d() | [16, 1, 224, 224]
| 5  | batch size, channels, depth, height, width | nn.Conv3d() | [16, 1, 3, 224, 224]


1.  A tensor [16] indicates there are 16 samples, each with a single label or
    prediction.
    
2. A tensor [16, 512] indicates there are 16 samples, each represented by a
   512-dimensional feature vector.

3. A tensor [16, 1, 512] indicates there are 16 samples, each with 1 channel
   (like grayscale in images or a single feature in sequences), and a sequence
   length of 512.

4.  A tensor [16, 1, 224, 224] indicates there are 16 images, each with 1
    channel (like grayscale), and a resolution of 224x224 pixels.

5. A tensor [16, 1, 3, 224, 224] indicates there are 16 samples, each with 1 channel, and dimensions of 3 (depth) by 224 (height) by 224 (width).

```python
import torch
import torch.nn as nn
import torchvision.transforms as transforms
from PIL import Image
from collections import OrderedDict
```

```python
input = torch.rand((1, 3, 32, 32))  # Batch Size, C, H, W
```

```python
model = nn.Sequential(OrderedDict([
    # 3 input channels, 8 output channels (will produce 8 feature maps), 3x3
    # filter
    ('conv1', nn.Conv2d(3, 8, 3)),

    # # introduces non-linearity to the model
    ('relu1', nn.ReLU()),
    
    # # size of the pooling window and stride
    ('pool', nn.MaxPool2d(2, 2)),  
    
    # number of input channels which matches the output channels from the
    # previous convolutional layer. Number of output channels will producie 16
    # feature maps. Size of the filter    
    ('conv2', nn.Conv2d(8, 16, 3)),
    
    # ReLU activation function.
    ('relu2', nn.ReLU()),
    
    # Max pooling layer with a size of 2 and stride of 2
    ('pool2', nn.MaxPool2d(2, 2)), 
    
    # Flattens the input, converts the multi-dimensional tensor into a 1D tensor
    ('flatten', nn.Flatten()),  
    
    # Number of input features which matches the flattened dimension of the
    # previous layer's output
    ('fc1', nn.Linear(16 * 6 * 6, 128)),

    # ReLU activation function 
    ('relu3', nn.ReLU()),

    # Number of input features which is the output from the previous layer.
    ('fc2', nn.Linear(128, 64)),

    # ReLU activation function
    ('relu4', nn.ReLU()),

    # The number of input features from the previous layer.
    ('fc3', nn.Linear(64, 1)),  

    # Output range between 0 and 1 (for binary classification tasks)
    ('sigmoid', nn.Sigmoid())  
]))
model(input).shape
```
