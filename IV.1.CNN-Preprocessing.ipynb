```python
from torchvision import transforms
from PIL import Image
```

```python
img = Image.open('raster_files/rgb_jpg.jpeg')
img
```

```python
img.size
```

```python
preprocess_steps = transforms.Compose([
    transforms.Resize((300, 300)),   # Resize to a square (300, 300)
    transforms.RandomRotation(50), # rotate image by 50 degrees
    transforms.CenterCrop(200), # Crop the center of the image
    # transforms.Grayscale(), # Convert the image to grayscale
    transforms.RandomVerticalFlip(), # Randomly flip the image vertically
    transforms.ToTensor(), # Convert the image to a PyTorch Tensor
transforms.Normalize((0.485, 0.456, 0.406), (0.229, 0.224, 0.225)),  # ImageNet values
])

x = preprocess_steps(img)
x
```

```python
x.mean([1, 2]), x.std([1, 2])
```
