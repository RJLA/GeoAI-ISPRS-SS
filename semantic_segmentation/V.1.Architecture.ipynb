# Understanding Feature Pyramid Network (FPN)

## What is FPN?

Feature Pyramid Network (FPN) is a neural network technique used in tasks like object detection and image segmentation. It helps the network detect objects of different sizes better by using information from multiple layers of a convolutional neural network (CNN).

## How Does FPN Work?


<img src="figures/fpn.png" alt="Alt text" width="600" height="400">


1. **Feature Maps**:
   - When an image passes through a CNN, it creates different "feature maps" at various layers.
   - **Shallow layers** (early in the network) capture small details like edges.
   - **Deep layers** (later in the network) capture more complex patterns like shapes or objects.

2. **Building the Pyramid**:
   - FPN takes feature maps from both shallow and deep layers.
   - It **upsamples** the deeper (low-resolution) maps to match the size of the shallower (high-resolution) maps.
   - These maps are then **combined** to create a richer, multi-scale feature map at each level.

3. **Multi-Scale Predictions**:
   - After combining, FPN can make predictions at different levels (or "scales") of the pyramid.
   - This means FPN is good at detecting objects of all sizes, from small to large.

## Why is FPN Useful?

- **Multi-Scale Detection**: FPN improves the network's ability to detect small, medium, and large objects by combining features from different layers.
- **Better Accuracy**: By using this pyramid of features, the network makes more accurate predictions.
