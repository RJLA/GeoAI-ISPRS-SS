```python
import os
import re
from pathlib import Path
from PIL import Image
import numpy as np
from patchify import patchify
```

```python
# Create the necessary folders for the semantic segmentation project (optional)
def create_folders(base_dir):
    """
    Creates the necessary folders for a semantic segmentation project.

    This function creates 'train', 'val', and 'test' folders inside the
    specified base directory.
    
    Each of these folders will contain 'images' and 'masks' subfolders.

    Args:
        base_dir (str): The base directory where the folders will be created.
    """

    FOLDERS = ['train', 'val', 'test']
    for folder in FOLDERS:
        folder_path = os.path.join(base_dir, folder)
        folder_imgs = os.path.join(folder_path, "images")
        folder_msks = os.path.join(folder_path, "masks")
        
        if not os.path.exists(folder_path):
            os.makedirs(folder_path)
        
        if not os.path.exists(folder_imgs):
            os.makedirs(folder_imgs)
        else:
            print(f'{folder_imgs} already exists')
        
        if not os.path.exists(folder_msks):
            os.makedirs(folder_msks)
        else:
            print(f'{folder_msks} already exists')

base_directory = os.getcwd()
create_folders(base_directory)
```

# High-Level Overview of the Patching

## Purpose
- **Generate image patches**: The script creates smaller patches from large images and masks for deep learning tasks like image segmentation.

- **Why Patching?**:
  - Facilitates handling of large images in deep learning.
  - Allows training on smaller, focused image areas, improving model performance.

## Summary
- The script efficiently breaks down large images into manageable patches, organizing them into datasets for training, validation, and testing in deep learning tasks.


```python
# Define the base directory
BASE_DIR = os.getcwd()

def create_patches(src, dest_path):
    """
    Create patches from an image and save them to the specified destination path.
    
    Args:
    - src (str): The path to the source image file.
    - dest_path (str): The destination path where the patches will be saved.
    """
    
    path_split = os.path.split(src)
    tile_num = re.findall(r'\d+', path_split[0])[0]
    
    image = Image.open(src)
    image = np.asarray(image)
    if len(image.shape) > 2:  # only if color channel exists as well
        patches = patchify(image, (320, 320, 3), step=300)
        file_name_wo_ext = Path(src).stem
        for i in range(patches.shape[0]):
            for j in range(patches.shape[1]):
                patch = patches[i, j, 0]
                patch = Image.fromarray(patch)
                num = i * patches.shape[1] + j
                patch.save(f"{dest_path}/{file_name_wo_ext}_tile_{tile_num}_patch_{num}.png")

# Loop through the data directory and create patches for each image and mask
data_dir = os.path.join(BASE_DIR, 'data')
for path_name, _, file_name in os.walk(data_dir):
    for f in file_name:
        print(f)
        if f != 'classes.json':
            
            path_split = os.path.split(path_name)
            tile_num = re.findall(r'\d+', path_split[0])[0]
            
            img_type = path_split[1]  # either 'masks' or 'images'
            
            # Define target folders
            if tile_num == '3':
                target_folder_imgs = os.path.join(BASE_DIR, 'val', 'images')
                target_folder_masks = os.path.join(BASE_DIR, 'val', 'masks')
            elif tile_num == '1':
                target_folder_imgs = os.path.join(BASE_DIR, 'test', 'images')
                target_folder_masks = os.path.join(BASE_DIR, 'test', 'masks')
            elif tile_num in ['4', '5', '6', '7', '8']:
                target_folder_imgs = os.path.join(BASE_DIR, 'train', 'images')
                target_folder_masks = os.path.join(BASE_DIR, 'train', 'masks')
            
            # copy all images
            src = os.path.join(path_name, f)
            file_name_wo_ext = Path(src).stem
            # check if file exists in images and masks
            img_file = os.path.join(path_split[0], 'images', f"{file_name_wo_ext}.jpg")
            mask_file = os.path.join(path_split[0], 'masks', f"{file_name_wo_ext}.png")
            if os.path.exists(img_file) and os.path.exists(mask_file):
                if img_type == 'images':
                    create_patches(src=src, dest_path=target_folder_imgs)
                    
                # copy all masks
                if img_type == 'masks':
                    create_patches(src=src, dest_path=target_folder_masks)
```
