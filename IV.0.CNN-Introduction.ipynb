## Convolutional Neural Networks (CNN) in Deep Learning

A **Convolutional Neural Network (CNN)** is a type of deep learning model designed for processing structured grid data, like images. It uses layers of filters (convolutions) to automatically learn features such as edges, textures, and patterns from raw input. CNNs are highly effective for image classification, object detection, and other tasks involving visual data.

### Key Components:
- **Convolutional Layers:** Extract features using filters.
- **Pooling Layers:** Reduce the spatial dimensions of features.
- **Fully Connected Layers:** Perform the final classification or prediction.

## Convolution Layer

The convolution layer is the core building block of the CNN. It carries the main portion of the network’s computational load.

This layer `performs a dot product between two matrices`, where one matrix is
the set of learnable parameters otherwise known as a kernel, and the other
matrix is the restricted portion of the receptive field. The kernel is spatially
smaller than an image but is more in-depth. This means that, if the image is
composed of three (RGB) channels, the kernel height and width will be spatially
small, but the depth extends up to all three channels.

## Pooling Layer

The **pooling layer** in a CNN reduces the spatial dimensions of the input, helping to down-sample features while retaining important information. This layer typically uses operations like **max pooling** or **average pooling** to aggregate values in small regions, making the network more efficient and reducing the risk of overfitting.
